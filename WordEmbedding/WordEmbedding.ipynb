Word Embedding / One-Hot Encoding

```python
S1 = '나는 책상 위에 사과를 먹었다'
S2 = '알고 보니 그 사과는 Jason 것이었다'
S3 = '그래서 Jason에게 사과를 했다'

token2idx={}
index=0

for sentence in [S1,S2,S3]:
    tokens =sentence.split()
    for token in tokens:
        if token2idx.get(token) == None:
            token2idx[token]=index
            index+=1
            
print(token2idx)
```

```python
V = len(token2idx)

token2vec = [([0 if i != idx else 1 for i in range(V)], idx, token) for token, idx in token2idx.items()]

for x in token2vec:
    print("\t".join([str(y) for y in x]))
```

```python
import numpy as np

for sentence in [S1,S2,S3]:
    onehot_s = []
    tokens = sentence.split()
    for token in tokens:
        if token2idx.get(token) != None:
            vector = np.zeros((1,V))
            vector[:,token2idx[token]] = 1
            onehot_s.append(vector)
        else: 
            print("UNK")
            
    print(f"{sentence}: ")
    print(np.concatenate(onehot_s, axis = 0))
    print("\n")
```

```python
# Bag of Word 예시
sentence = ["그래서", "Jason에게", "사과를", "했다"]

# 단어 등장 횟수 세기
freq_dict = {}
for token in sentence:
    if token in freq_dict:
        freq_dict[token] += 1
    else:
        freq_dict[token] = 1

print(freq_dict)
```

```python
paragraph = ["나는", "사과를", "먹었다", "그래서", "Jason에게", "사과를", "했다"]

freq_dict = {}
for token in paragraph:
    freq_dict[token] = freq_dict.get(token, 0) + 1

print(freq_dict)
```

```python
import torch
from datasets import load_dataset
from transformers import AutoTokenizer

# 데이터 로드
dataset = load_dataset("imdb")

# 전처리 (기존 코드의 fix_length=500과 동일)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

def tokenize_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True, max_length=500)

tokenized_datasets = dataset.map(tokenize_function, batched=True)
tokenized_datasets.set_format(type='torch', columns=['input_ids', 'label'])

print("복구된 2.8.0 환경에서 데이터 로드 완료.")
```

```python
# 'appropriate update'라는 단어가 들어간 리뷰 검색
search_term = "appropriate update"
results = [d for d in dataset['train'] if search_term in d['text'].lower()]

print(f"\n검색 결과: {len(results)}개의 리뷰를 찾았습니다.")

if results:
    print("-" * 50)
    print("첫 번째 검색 결과 내용:")
    print(results[0]['text'][:500] + "...") # 500자만 출력
```

```python
# 훈련 데이터셋의 첫 번째 샘플 꺼내기
sample = tokenized_datasets['train'][0]

print("--- 데이터 샘플 확인 ---")
print(f"라벨 (0:부정, 1:긍정): {sample['label']}")
print(f"숫자로 변환된 텍스트(ID) 일부: {sample['input_ids'][:10]}") 
print(f"전체 길이: {len(sample['input_ids'])}") # 설정하신 500이 나와야 합니다.
```
