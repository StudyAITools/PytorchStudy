Word Embedding / One-Hot Encoding

```python
S1 = '나는 책상 위에 사과를 먹었다'
S2 = '알고 보니 그 사과는 Jason 것이었다'
S3 = '그래서 Jason에게 사과를 했다'

token2idx={}
index=0

for sentence in [S1,S2,S3]:
    tokens =sentence.split()
    for token in tokens:
        if token2idx.get(token) == None:
            token2idx[token]=index
            index+=1
            
print(token2idx)
```

```python
V = len(token2idx)

token2vec = [([0 if i != idx else 1 for i in range(V)], idx, token) for token, idx in token2idx.items()]

for x in token2vec:
    print("\t".join([str(y) for y in x]))
```

```python
import numpy as np

for sentence in [S1,S2,S3]:
    onehot_s = []
    tokens = sentence.split()
    for token in tokens:
        if token2idx.get(token) != None:
            vector = np.zeros((1,V))
            vector[:,token2idx[token]] = 1
            onehot_s.append(vector)
        else: 
            print("UNK")
            
    print(f"{sentence}: ")
    print(np.concatenate(onehot_s, axis = 0))
    print("\n")
```

```python
# Bag of Word 예시
sentence = ["그래서", "Jason에게", "사과를", "했다"]

# 단어 등장 횟수 세기
freq_dict = {}
for token in sentence:
    if token in freq_dict:
        freq_dict[token] += 1
    else:
        freq_dict[token] = 1

print(freq_dict)
```

```python
paragraph = ["나는", "사과를", "먹었다", "그래서", "Jason에게", "사과를", "했다"]

freq_dict = {}
for token in paragraph:
    freq_dict[token] = freq_dict.get(token, 0) + 1

print(freq_dict)
```

Reading Data

```python
from datasets import load_dataset
from transformers import AutoTokenizer

# 데이터 로드 (Hugging Face 라이브러리 사용)
raw_datasets = load_dataset("imdb")

# 토크나이저 준비 (BERT 기반 토크나이저가 현재 업계 표준임)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# 설정값 보관
fix_length = 500

# Data Length 확인
print(f"Train Data Length : {len(raw_datasets['train'])}")
print(f"Test Data Length : {len(raw_datasets['test'])}")

# Data Fields 확인 (Label과 Text 컬럼 확인)
print(raw_datasets['train'].column_names)

# Data Sample 확인
print('---- Data Sample ----')
print('Input : ')
print(raw_datasets['train'][1]['text'], '\n')
print('Label : ')
print(raw_datasets['train'][1]['label']) # 0: neg, 1: pos
```

Pre-processing Data

```python
# Transformers는 토크나이저 내부에서 소문자화, 특수문자 처리를 지원함
# fix_length=500과 패딩/커팅을 한 번에 처리하는 함수 정의
def tokenize_function(examples):
    return tokenizer(
        examples["text"], 
        padding="max_length", 
        truncation=True, 
        max_length=fix_length
    )

# 데이터셋 전체에 실시간 전처리 적용 (map 방식이 루프보다 압도적으로 빠름)
tokenized_datasets = raw_datasets.map(tokenize_function, batched=True)

# 전처리 후 결과 샘플 확인 (수치화된 ID 확인)
print('---- Preprocessed Data Sample (Input IDs) ----')
print(tokenized_datasets['train'][1]['input_ids'][:20]) # 처음 20개 토큰만 출력
```

Making Vocab & Setting Embedding

```python
# 이미 토크나이저 안에 수만 개의 단어가 포함되어 있음
vocab_size = tokenizer.vocab_size
model_config = {'vocab_size': vocab_size, 'emb_dim': 768} # BERT 기본 차원

# Vocabulary Info 확인 (stoi 역할)
print(f'Vocab Size : {vocab_size}')
print('Vocab Examples : ')
# 10개만 추출해서 확인
for word, idx in list(tokenizer.get_vocab().items())[:10]:
    print('\t', word, idx)

print('---------------------------------')

# Label Info 확인 (IMDB는 0: neg, 1: pos로 고정됨)
label_names = raw_datasets['train'].features['label'].names
print(f'Label Size : {len(label_names)}')
print('Lable Examples : ')
for i, name in enumerate(label_names):
    print('\t', name, i)
```

Spliting Validation Data & Making Data Iterator

```python
import torch
from torch.utils.data import DataLoader

# 훈련 데이터를 8:2로 다시 분할
train_valid_split = tokenized_datasets["train"].train_test_split(test_size=0.2, seed=0)
train_ds = train_valid_split["train"]
valid_ds = train_valid_split["test"]
test_ds = tokenized_datasets["test"]

# PyTorch용 텐서 포맷으로 변경
train_ds.set_format(type='torch', columns=['input_ids', 'label'])
valid_ds.set_format(type='torch', columns=['input_ids', 'label'])
test_ds.set_format(type='torch', columns=['input_ids', 'label'])

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# DataLoader 생성 (기존 BucketIterator 역할)
train_loader = DataLoader(train_ds, batch_size=30, shuffle=True)
valid_loader = DataLoader(valid_ds, batch_size=30)
test_loader = DataLoader(test_ds, batch_size=30)
```

5. Check Batch Data

```python
# 배치 데이터 하나 확인
batch = next(iter(train_loader))

# 텐서 형태 출력
print(f"Batch Size/Fix Length: {batch['input_ids'].shape}") # [30, 500]
print(f"Labels: {batch['label']}")

# 역변환 확인 (숫자 -> 문장)
# tokenizer.decode를 사용하면 패딩 등을 제외하고 깔끔하게 문장으로 복구함
decoded_sentence = tokenizer.decode(batch['input_ids'][0], skip_special_tokens=True)
print(f"Reverted Sentence: {decoded_sentence[:200]}...")

# 라벨 정보 확인
print(f"Label Mapping: 1 is 'pos', 0 is 'neg' -> Result: {batch['label'][0]}")
```
